# Exercise 2: Q-Learning with FrozenLake

In this exercise, we will implement our first real learning algorithm from scratch: Q-Learning. We will build a "quality table" (Q-table) that will guide the agent's decisions to solve the `FrozenLake-v1` environment.

```python
import logging
import sys
import os
import gymnasium as gym

# Configure logging
logging.basicConfig(level=logging.INFO, format='%(message)s', force=True)
sys.path.append(os.path.abspath('..'))

from src.exercise_2.qtable import initialize_q_table
from src.exercise_2.training import train_q_learning
from src.exercise_2.evaluation import evaluate_agent
from src.exercise_2.visualization import display_trained_agent_interaction
```

## Step 1: Environment and Q-Table Initialization

We start by creating the `FrozenLake-v1` environment. By default, FrozenLake is slippery, making it a challenging environment.

Then, we initialize our Q-table as a NumPy array filled with zeros, with one row for each state and one column for each action.

```python
env_id = 'FrozenLake-v1'
# We use the default environment (is_slippery=True by default)
env = gym.make(env_id)

q_table = initialize_q_table(env)
print(f"Q-Table Dimensions: {q_table.shape}")
print("Initial Q-Table preview:\n", q_table)
```

## Step 2: Agent Training

At each step, the agent decides whether to explore or exploit (epsilon-greedy strategy). Then, it updates the Q-Table using the Bellman equation. 

We will train the agent for 10,000 episodes to give it time to map out the best routes on the slippery ice.

```python
q_table = train_q_learning(env, q_table, num_episodes=10000)
```

## Step 3: Evaluation

Now that the learning phase is over, we will test the agent over 100 episodes. This time, the agent no longer explores: it strictly follows the optimal policy found in its Q-Table.

```python
success_rate = evaluate_agent(env, q_table, num_episodes=100)
```

## Step 4: Visualization

Finally, let's watch our trained agent navigate the frozen lake! The animation below uses the fully learned `Q-table` to make decisions at every step.

```python
# Play a single episode to visualize the agent's learned policy
display_trained_agent_interaction(env_id, q_table, max_steps=100)
```
